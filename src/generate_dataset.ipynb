```python
from tqdm import trange
import torch
```

```python
train_dataset_size = 10_000

samples = []
for i in trange(train_dataset_size):
    img, labels = generate_image_and_labels()
    samples.append({"image": img, "labels": labels})

torch.save(samples, "/tmp/train_dataset.pt")
```

```python
val_dataset_size = 1_000

samples = []
for i in trange(val_dataset_size):
    img, labels = generate_image_and_labels()
    samples.append({"image": img, "labels": labels})

torch.save(samples, "/tmp/val_dataset.pt")
```
